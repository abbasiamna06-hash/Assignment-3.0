```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect("database.sqlite")
```

```python
df = pd.read_sql_query("SELECT * FROM Salaries", conn)
```

```python
df.shape
```

```python
df.columns
```

## **What is the average TotalPayBenefits for each JobTitle? exclude ('Not Provided', 'Not provided')**

```python
df = df[df["JobTitle"] != "Not Provided"]
df = df[df["JobTitle"] != "Not provided"]

Avg = df.groupby("JobTitle")["TotalPayBenefits"].mean()

Avg
```

## **Show The JobTitle and Overtime Pay for All Employees With Overtime Pay greater than 50000**

```python
df["OvertimePay"] = pd.to_numeric(df["OvertimePay"])
```

```python
Pay = df[df["OvertimePay"] > 50000][["JobTitle", "OvertimePay"]]

Pay
```

## **What is the total number of unique employees? exclude ('Not Provided', 'Not provided')**

```python
df = df[df["EmployeeName"] != "Not Provided"]
df = df[df["EmployeeName"] != "Not provided"]

unique_employees = df["EmployeeName"].nunique()

unique_employees
```

## **Show The Top 10 highest Paid Employees**

```python
Top_10 = df[
    ["EmployeeName", "JobTitle", "TotalPayBenefits"]
].sort_values(
    "TotalPayBenefits",
    ascending=False
).head(10)

Top_10 
```

## **Show All Employees Ordered By Their TotalPay Benefits In Descending Order**

```python
Employees = df[
    ["EmployeeName", "JobTitle", "TotalPayBenefits"]
].sort_values(
    "TotalPayBenefits",
    ascending=False
)

Employees
```
