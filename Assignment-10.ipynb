```python
import pandas as pd

file = open("retail supermarket - retail supermarket.csv", "r")

df = pd.read_csv(file)

file.close()


result = df[(df["Sales"] > 500) & (df["Quantity"] > 5)]
print(result)
```

```python
result = df[(df["Discount"] > 0.5) | (df["Profit"] < 0)]
print(result)
```

```python
result = df.loc[df["Region"] == "East", ["Ship Mode", "Profit"]]
print(result)
```

```python
result = df[(df["Segment"] == "Corporate") & 
            (df["Region"] == "Central")]
print("Number of records:", len(result))
```

```python
result = df[(df["Category"] == "Furniture") & 
            (df["Sub-Category"] != "Chairs")]
print(result)
```

```python
result = df[(df["Discount"] > 0.3) & 
            (df["Profit"] > 0)]
print(result)
```

```python
result = df[(df["Quantity"] > 5) & 
            (df["Category"] == "Technology") & 
            (df["Profit"] > 0)]
print(result)
```

```python
result = df.iloc[0:5, [5, 6, 11]]
print(result)
```

```python
result = df[(df["Segment"] == "Home Office") & 
            (df["Sales"] > 1000) & 
            (df["Quantity"] > 3)]
print(result)
```

```python
df["Profit Margin"] = df["Profit"] / df["Sales"]

result = df[df["Profit Margin"] > 0.3]
print(result)
```
