```python
import pandas as pd
```

```python
df = pd.read_csv('cars.csv')
df
```

```python
print(df.duplicated())
```

```python
df = df.drop_duplicates()
df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df
```

```python
result = df.groupby("Make")["MSRP"].mean()
print(result.head(10).round(2))
```

```python
result = df.groupby("Year")["MSRP"].mean()
print(result.round(2))
```

```python
result = df.groupby("Year")[["Engine HP", "city mpg", "highway MPG"]].mean()

result["MPG"] = (result["city mpg"] + result["highway MPG"]) / 2

result = result[["Engine HP", "MPG"]]

result.round(2)
```

```python
size_cars = df[df["Vehicle Size"].isin(["Compact", "Midsize", "Large"])]

percentage = (len(size_cars) / len(df)) * 100

print("Percentage:", round(percentage, 2), "%")
```
