Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.

```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("\nEmployee Information:")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)

    def update_year_of_experience(self, new_experience):
        self.year_of_experience = new_experience
        print("Year of Experience updated to:", self.year_of_experience)

    def increase_salary(self, increment):
        self.salary = self.salary + increment
        print("Salary increased to:", self.salary)


employee101 = Employee(
    "Ali",
    101,
    "Karachi",
    50000,
    "Software Engineer",
    2022,
    "BS Computer Science",
    3
)

print("Name:", employee101.name)
print("Number:", employee101.number)
print("Address:", employee101.address)
print("Salary:", employee101.salary)

employee101.describe_employee()

employee101.increase_salary(10000)

employee101.update_year_of_experience(4)


employee1 = Employee("Ahmed", 102, "Lahore", 45000, "Developer", 2023, "BSCS", 2)
employee2 = Employee("Sara", 103, "Karachi", 60000, "Manager", 2020, "MBA", 6)
employee3 = Employee("Hassan", 104, "Islamabad", 55000, "Designer", 2021, "BSIT", 5)
employee4 = Employee("Ayesha", 105, "Multan", 40000, "HR Officer", 2024, "BBA", 1)
employee5 = Employee("Usman", 106, "Peshawar", 70000, "Team Lead", 2019, "BSCS", 7)
employee6 = Employee("Fatima", 107, "Quetta", 50000, "Accountant", 2022, "B.Com", 3)
employee7 = Employee("Hamza", 108, "Faisalabad", 48000, "Web Developer", 2022, "BSCS", 4)
employee8 = Employee("Zainab", 109, "Rawalpindi", 65000, "Data Analyst", 2020, "BSIT", 6)
employee9 = Employee("Bilal", 110, "Sialkot", 52000, "Software Engineer", 2021, "BSCS", 5)
employee10 = Employee("Hira", 111, "Hyderabad", 58000, "Project Manager", 2019, "MBA", 7)


employees = [
    employee1, employee2, employee3, employee4, employee5,
    employee6, employee7, employee8, employee9, employee10
]

for employee in employees:
    employee.describe_employee()

employee1.increase_salary(5000)
employee2.update_year_of_experience(7)
employee3.increase_salary(8000)
employee4.update_year_of_experience(2)
employee5.increase_salary(10000)
```
