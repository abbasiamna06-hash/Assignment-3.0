```python
import csv

open_file = open("student data - student data.csv", "r")
open_file

reader = csv.reader(open_file)
reader

data = list(reader)
data

header = data[0]
header
header.append("Grade")
print(header)

student_data = data[1:]
student_data

for per in student_data:
    percentage = float(per[5])

    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    per.append(grade)
    print(per)

open_write = open("data.csv", mode="w")
open_write

writer = csv.writer(open_write)
writer

writer.writerows(data)

open_write.close()
```
